```python
X = df.drop('price', axis=1)
y = df['price']
```

```python
X
```

```python
y
```

```python
X_onehot_encode= pd.get_dummies(X,columns = ['fuelType','transmission','model'], drop_first=True)
X_onehot_encode
```

```python
X_onehot_encode = X_onehot_encode.astype(int)
X_onehot_encode
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
cols = ['fuelType','transmission','model']


Xlabel = X.copy()
for i in cols:
    Label_encode = LabelEncoder()
    Xlabel[i] = Label_encode.fit_transform(Xlabel[i])
    
Xlabel
    
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
num_cols = ['year', 'mileage', 'engineSize','tax','mpg']

scaler = StandardScaler()
X_onehot_encode[num_cols] = scaler.fit_transform(X_onehot_encode[num_cols])

X_onehot_encode
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
train_X, test_X, train_y, test_y = train_test_split(X_onehot_encode, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(train_X, train_y)
```

```python
y_pred = model.predict(test_X)
```

```python
r2 = r2_score(test_y, y_pred)
r2
```

```python
n = test_X.shape[0]
p = test_X.shape[1]

n,p
```

```python
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adjusted_r2
```
